## 1. Introducción

El objetivo de esta práctica es desarrollar una aplicación web capaz de clasificar imágenes mediante un modelo de Inteligencia Artificial. Para ello, se utiliza Google Teachable Machine, una herramienta que permite entrenar modelos de aprendizaje automático sin necesidad de programarlos desde cero.

Posteriormente, el modelo se integra en una aplicación desarrollada con HTML, CSS y JavaScript mediante TensorFlow.js.

## 2. Creación y entrenamiento del modelo

En primer lugar, se accede a Google Teachable Machine y se crea un proyecto de tipo imagen utilizando el modelo estándar.

Se definen dos categorías de imágenes y se recopilan muestras para cada una. Estas imágenes sirven para que el modelo aprenda a diferenciar entre las clases seleccionadas.

Una vez recogidas las muestras, se inicia el entrenamiento y se realizan pruebas para comprobar si el modelo clasifica correctamente las imágenes.

## 4. Desarrollo de la aplicación web

Se desarrolla una aplicación web utilizando HTML para estructurar el contenido, CSS para diseñar la interfaz y JavaScript para implementar la funcionalidad.

La aplicación carga el modelo entrenado y permite realizar predicciones sobre las imágenes seleccionadas por el usuario.

## 5. Funcionalidades implementadas

Se incorporan varias mejoras para ampliar las posibilidades de la aplicación:

- **Diseño responsive:** la interfaz se adapta a diferentes tamaños de pantalla.
- **Control de cámara:** permite iniciar y detener la webcam para clasificar imágenes en tiempo real.
- **Mensajes personalizados:** muestra el resultado de la clasificación y la confianza obtenida.
- **Umbral de confianza:** permite establecer un porcentaje mínimo para considerar válida una coincidencia.
- **Historial de resultados:** guarda y muestra las últimas clasificaciones realizadas.
- **Subida de imágenes:** permite seleccionar imágenes almacenadas en el dispositivo y analizarlas con el modelo.

El historial se almacena en el navegador mediante `localStorage`, por lo que los resultados pueden mantenerse al cerrar o recargar la página en el mismo navegador.
